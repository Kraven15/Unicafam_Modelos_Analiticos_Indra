```python
# Ejercicio 1: Clasificador de notas

notas = [float(input('Ingresa la nota: '))]

for n in notas:
  if n < 3.0:
    print('Reprueba')
  elif 3.0 <= n <= 4.4:
    print('Aprueba')
  else:
    print('Excelente')

```

```python
# Ejercicio 2: Carrito de compras

carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

subtotal = 0
for producto, (precio, cantidad) in carrito.items():
  subtotal += precio * cantidad


if subtotal > 200000:
  subtotal -= subtotal * 0.10


total = subtotal * 1.19
print(f"Total a pagar: ${total:,.0f}".replace(",", "."))
```

```python
# Ejercicio 3: Nómina con pandas

import numpy as np
import pandas as pd

empleados = pd.DataFrame({
    "nombre": ["Pedro", "Juan", "Camilo"],
    "area": ["soporte", "desarrollo", "pruebas"],
    "salario": [2000000, 6500000, 5500000]
})

```

```python
# Agregar 3 empleados nuevos
nuevos_empleados = pd.DataFrame({
    'nombre': ['Marta', 'Emili', 'Sofía'],
    'area': ['Ventas', 'TI', 'Ventas'],
    'salario': [2800000, 4200000, 2600000]
})

empleados = pd.concat([empleados, nuevos_empleados], ignore_index=True)

print(empleados)
```

```python
# Crea la columna bono: 10%

empleados["bono"] = empleados["salario"] * 0.05

empleados.loc[empleados["area"] == "Ventas", "bono"] = empleados["salario"] * 0.10

print(empleados)
```

```python
# costo total de nómina mensual por área

empleados["nomina"] = empleados["salario"] + empleados["bono"]

print(empleados.groupby("area")["nomina"].sum())
```

```python
# empleado con mayor salario
print(empleados.loc[empleados["salario"].idxmax()])
```
