```python
# Ejercicio 3: Primer análisis de ventas

# Creacion de tabla
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200

ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})


```

```python
# total vendido por canal
ventas["total"] = ventas["unidades"] * ventas["precio"]

total_canal = ventas.groupby("canal")["total"].sum()

print(total_canal)
```

```python
# ticket promedio por ciudad

ticket = ventas.groupby("ciudad")["total"].mean()

print(ticket.round(2))
```

```python
# día de la semana se vende más
ventas["dia_semana"] = ventas["fecha"].dt.day_name()

ventas_por_dia = ventas.groupby("dia_semana")["total"].sum()

ventas_por_dia = ventas_por_dia.sort_values(ascending=False)

print(ventas_por_dia)
```
